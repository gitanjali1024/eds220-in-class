```python
import numpy as np
import pandas as pd

# Load Palmer penguins data
URL = 'https://raw.githubusercontent.com/allisonhorst/palmerpenguins/main/inst/extdata/penguins.csv'
penguins = pd.read_csv(URL)

penguins.head()
```

```python
# Get the number of non-NA values in each column
penguins.count()
```

```python
# Get minimum value in each column with numerical values
penguins.select_dtypes('number').min()
```

```python
penguins['flipper_length_mm'].mean()
```

```python
penguins.groupby('species')['flipper_length_mm']
```

```python
# Average flipper length per species
penguins.groupby('species')['flipper_length_mm'].mean()
```

```python
# Average flipper length per species
avg_flipper = (penguins.groupby('species')['flipper_length_mm'].mean()
              .rename('mean_flipper_length')
              .sort_values(ascending=False))
avg_flipper
```

```python
# Average flipper length per species and island
penguins.groupby(['species','island'])['flipper_length_mm'].mean()
```

```python
# Number of NA observations
penguins['sex'].size - penguins['sex'].count()
```

```python
penguins.groupby('sex').size()
```

```python
penguins.groupby('sex', dropna = False).size()
```

# Check In

```python
penguins.groupby(['island','year']).count()
```

```python
penguins.groupby(['island', 'year']).size()
```

.count() gives us the count of every variable within the grouping of 'island' and 'year', so columns with NA values will have a smaller count than those without. .size() simply gives us how many rows are in the table, which is the same for every variable. This is more useful for giving us the number of penguins surveyed.

```python
(penguins.groupby(['island','year'])
.size()
.sort_values()
.plot(kind = 'barh',
     title = 'Penguins surveyed at the Palmer Archipelago',
     xlabel = 'Number of penguins',
     ylabel = 'Island, Year'))
```

```python
penguins.groupby(['year', 'species'])['body_mass_g'].max()
```

```python
(penguins.groupby(['year', 'species'])['body_mass_g'].max()
.sort_values()
.plot(kind = 'barh',
             title = 'Maximum penguin body mass by Species and Island',
             xlabel = 'Body Mass (g)',
             ylabel = 'Species, Year'))
```
